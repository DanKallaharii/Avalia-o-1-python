```python
saldo = 200

def depositar():
  global saldo
  deposito = float(input("Digite o valor do Deposito:\n"))
  if deposito <= 0:
    print("Deposito Invalido")
  else:
    print(f"Seu deposito de {deposito:.2f}R$ Reais foi realizado com sucesso\n")
    saldo += deposito

def sacar():
  global saldo
  saque = float(input("Digite o valor do Saque\n"))
  if saque > saldo:
    print("Saldo Insuficiente\n")
  elif saque == 0:
    print("Saque Invalido")
  else:
    print(f"Seu saque de {saque:.2f}R$ Reais foi realizado com sucesso\n")
    saldo -= saque

def consultar_saldo():
  print(f"Seu Saldo atual é de {saldo:.2f}")
def sair():
  print("Saindo...")

while True:
  print("[1] Depositar\n[2] Sacar\n[3] Consultar Saldo\n[4] Sair\n")
  op = int(input("Escolha uma opção:\n"))
  if op == 1:
    depositar()
  elif op == 2:
    sacar()
  elif op == 3:
    consultar_saldo()
  elif op == 4:
    sair()
    break
  else:
    print("Opção Invalida\n")
```
