```python
#Declaração da classe
class Pessoa:
  def __init__(self): #método construtor
    self.nome = "" #atributdos de instancia
    self.idade = 0


    #metados de intancia
  def aniversario(self):
      self.idade += 1

  def menssagem(self):
      return f"{self.nome} é uma pessoa e tem {self.idade} anos \n"
#declaração do obejeto

pessoa1 = Pessoa()
pessoa1.nome = "Daniel "
pessoa1.idade = 26 #isso é um atributo
pessoa1.aniversario() #isso é um método
print(pessoa1.menssagem())

```
