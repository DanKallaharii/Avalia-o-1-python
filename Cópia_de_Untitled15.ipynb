```python

class ContaBancaria:
  ''' cria uma conta bancaria'''
  def __init__(self,id,nome,saldo = 0):
    self.id = id
    self.titular = nome
    self.saldo = saldo
  def __str__(self):
     return f"A conta {self.id} de {self.titular}, tem R${self.saldo:.2f} reais"

  def depositar(self,valor):
    self.saldo += valor
    print(f"Deposito de {valor:.2f} autorizado na conta de {self.id} ")


  def sacar(self,valor):
    if valor > self.saldo:
      print("Saque negado")
    else:
      self.saldo -= valor
      print(f"saque de {valor:.2f} autorizado na conta de {self.id} ")




conta1 = ContaBancaria(114,"Daniel",4.97)

conta1.depositar(100)
conta1.sacar(50)
print(conta1)

```
